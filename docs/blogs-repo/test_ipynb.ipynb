```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Import Required Libraries

# Simple Plotting Notebook

This notebook demonstrates basic plotting with Matplotlib, NumPy, and Pandas.

## Create Sample Data

```python
# Generate sample data
x = np.linspace(0, 10, 50)
y = np.sin(x)
y_scatter = y + np.random.normal(0, 0.1, len(x))

# Create a DataFrame
df = pd.DataFrame({'x': x, 'y': y, 'y_scatter': y_scatter})
```

## Create a Basic Line Plot

```python
plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True, alpha=0.3)
plt.show()
```

## Create a Scatter Plot

```python
plt.figure(figsize=(10, 6))
plt.scatter(x, y_scatter, alpha=0.6, s=50)
plt.title('Scatter Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True, alpha=0.3)
plt.show()
```

## Customize Plot Appearance

```python
plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2, label='Sin(x)', color='blue', marker='o', markersize=4)
plt.scatter(x, y_scatter, alpha=0.5, s=50, label='Noisy Data', color='red')
plt.title('Customized Plot with Multiple Elements', fontsize=14, fontweight='bold')
plt.xlabel('X-axis', fontsize=12)
plt.ylabel('Y-axis', fontsize=12)
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3, linestyle='--')
plt.show()
```

## Display Multiple Plots

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Line plot
axes[0, 0].plot(x, y, color='blue', linewidth=2)
axes[0, 0].set_title('Line Plot')
axes[0, 0].set_xlabel('X')
axes[0, 0].set_ylabel('Y')
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Scatter plot
axes[0, 1].scatter(x, y_scatter, color='red', alpha=0.6)
axes[0, 1].set_title('Scatter Plot')
axes[0, 1].set_xlabel('X')
axes[0, 1].set_ylabel('Y')
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Bar plot
axes[1, 0].bar(x[::5], y[::5], color='green', alpha=0.7, width=0.5)
axes[1, 0].set_title('Bar Plot')
axes[1, 0].set_xlabel('X')
axes[1, 0].set_ylabel('Y')
axes[1, 0].grid(True, alpha=0.3, axis='y')

# Plot 4: Histogram
axes[1, 1].hist(y_scatter, bins=20, color='orange', alpha=0.7, edgecolor='black')
axes[1, 1].set_title('Histogram')
axes[1, 1].set_xlabel('Values')
axes[1, 1].set_ylabel('Frequency')
axes[1, 1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```
